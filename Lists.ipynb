```python
# Enter your code here. Read input from STDIN. Print output to STDOUT

from collections import deque 
d = deque()

N = int(input())
comm = []

for i in range(1, N+1):
    comm.append(input())

comm = [x.split(' ') for x in comm]

for x in comm:
    for i in range(1, len(x)):
        x[i] = int(x[i])

for x in comm:
    match x[0]:
        case 'append': d.append(x[1])
        
        case 'appendleft': d.appendleft(x[1])
        
        case 'pop': d.pop()
        
        case 'popleft': d.popleft()
        
print(*d)
                        
```
